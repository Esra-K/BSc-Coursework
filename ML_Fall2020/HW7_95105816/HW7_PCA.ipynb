```
import numpy as np
import matplotlib.pyplot as plt
trn = np.load("train.npy")
tst = np.load("test.npy")
a = trn.shape[2]
b = trn.shape[3]
# n = 5
trn = trn.reshape((trn.shape[0] * trn.shape[1], trn.shape[2] * trn.shape[3]))
tst = tst.reshape((tst.shape[0] * tst.shape[1], tst.shape[2] * tst.shape[3]))
trn = (trn.T - np.mean(trn, axis=1)).T
tst = (tst.T - np.mean(tst, axis=1)).T

cova = np.cov(trn.T, rowvar=False, bias=True)
eigen_values, eigen_vectors = np.linalg.eig(cova)
print(eigen_vectors.shape)
print(cova.shape)
print(trn.shape)
print(tst.shape)

n = 5
projection_matrix = (eigen_vectors.T[:][:n])
print(projection_matrix)
eigenpca = np.matmul(trn.T, projection_matrix.T)
print(eigenpca.shape)
eigenpca = np.transpose(eigenpca)
eigenpca = eigenpca.reshape((eigenpca.shape[0], a, b))
for i in range(eigenpca.shape[0]):
    plt.imshow(eigenpca[i], interpolation='nearest')
    plt.show()
```
